## Task 2
Are there any null values in the Grc_df? If yes, then in which columns and how many? Finally, handle these null values using any strategy shown during the labs.

```python
Grc_df.isna().sum()
```

```python
Grc_df["Outlet_Size"].value_counts()
```

```python
Grc_df["Item_Weight"].describe()
```

```python
Grc_df["Outlet_Size"] = Grc_df["Outlet_Size"].fillna(
    Grc_df["Outlet_Size"].mode()[0]
)

Grc_df["Item_Weight"] = Grc_df["Item_Weight"].fillna(

    Grc_df["Item_Weight"].median()
)

Grc_df.isna().sum()
```

There are 818 null values in Item_Weight and 1439 null values in Outlet_Size. Since Outlet_Size is categorical, the missing values were filled with the mode. For the numerical Item_Weight column, the missing values were filled with the median. After handling the missing values, there are no null values remaining in the dataset.

## Task 7 
Using Seaborn (“ggplot style”) create a (2,1) subplot of a box plot showing 5-point summary of the column Item_MRP for SupType_1 and SupType_2. Which Outlet Type has a higher median MRP? Are there any outliers?

```python
plt.style.use("ggplot")
fig, axes = plt.subplots(2, 1)

sns.boxplot(x=SupType_1["Item_MRP"], ax=axes[0])
sns.boxplot(x=SupType_2["Item_MRP"], ax=axes[1])

axes[0].set_title("Supermarket Type1")
axes[1].set_title("Supermarket Type2")

plt.tight_layout()
plt.show()
```

```python
SupType_1["Item_MRP"].median(), SupType_2["Item_MRP"].median()
```

Supermarket Type1 has a higher median Item_MRP (144.75) compared to Supermarket Type2 (130.18). Based on the box plots, there are no outliers in either supermarket type.
